# Sales Data Analysis

This notebook analyzes sales data to understand revenue trends across months, regions, and products.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales_data.csv")
df.head()
```

```python
df["Date"] = pd.to_datetime(df["Date"])
df["Revenue"] = df["Units_Sold"] * df["Price"] * (1 - df["Discount"])
```

## Monthly Revenue Analysis

```python
monthly_revenue = df.groupby(df["Date"].dt.month_name())["Revenue"].sum()
monthly_revenue
```

```python
monthly_revenue.plot(kind="bar")
plt.show()
```

April generated the highest revenue, contributing the largest share of total sales.
